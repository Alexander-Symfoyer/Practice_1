**Find the number of Action squares with MERGE**

```python
movies_df = pd.read_csv('/content/drive/MyDrive/IMDb movies.csv')
action_movie_df = movies_df[movies_df['genre']
                            .apply(lambda m : 'Action' in m)]
```

```python
title_principals_df = pd.read_csv('/content/drive/MyDrive/IMDb title_principals.csv')
title_principals_df.head()
```

```python
merged_df = action_movie_df.merge(title_principals_df,
                      how='inner' ,
                      left_on = 'imdb_title_id' ,
                      right_on = 'imdb_title_id')
```

```python
action_star_df = merged_df[merged_df['category'].isin(['actor','actress'])]
```

```python
names_df = pd.read_csv('/content/drive/MyDrive/IMDb names.csv')
action_star_df.merge(names_df ,
                     how = 'inner' ,
                     left_on = 'imdb_name_id' ,
                     right_on = 'imdb_name_id')[
                         ['imdb_name_id','name']
                         ].drop_duplicates()
```

**Filter Data by YEAR**

```python
movies_df = pd.read_csv('/content/drive/MyDrive/IMDb movies.csv',
                        dtype = {'year' : int})
```

```python
new_movies_df = movies_df[movies_df['year'] >= 2015]
new_movies_df.head(5)
```

**Choose the most PROFITABLE Data**

```python
filtered_movies_df = new_movies_df[new_movies_df['budget'].notna() &
              new_movies_df['worlwide_gross_income'].notna()
              ][
                  ['imdb_title_id',
                   'budget',
                   'worlwide_gross_income'
                  ]
              ]
```

```python
filtered_movies_df['worlwide_gross_income'].apply(lambda m : m.split()[0]).unique()
```

```python
filtered_movies_df['budget'].apply(lambda m : m.split()[0]).unique()
```

```python
filtered_movies_df = filtered_movies_df[
    filtered_movies_df['budget'].str.slice(stop=1) == '$']
```

```python
filtered_movies_df['profit'] = filtered_movies_df['worlwide_gross_income'].apply(lambda m : int(m.split()[1])) - filtered_movies_df['budget'].apply(lambda m : int(m.split()[1]))
filtered_movies_df
```

```python
selected_ids = filtered_movies_df[filtered_movies_df['profit'] > 0 ]['imdb_title_id']
selected_ids
```

**Find AUDIENCE Group**

```python
popular_movie_df = movies_df.merge(selected_ids)
popular_movie_df['genre'] = popular_movie_df['genre'].apply(lambda m: m.split(', '))
popular_movie_df = popular_movie_df.explode('genre')
popular_movie_df.groupby('genre')['votes'].sum()
```
