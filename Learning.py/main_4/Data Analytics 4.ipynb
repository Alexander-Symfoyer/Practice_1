**Plotting**

```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
x,y = [x for x in range(10)], np.random.randint(0,10, size=10)
print('x : {}'.format(x))
print('y : {}'.format(y))
```

```python
freq = [np.random.randint(0,10, size=100)]
freq
```

```python
fig = plt.figure(figsize = (10,5))

ax_line = fig.add_subplot(2,3,1)
ax_line.plot(x,y)

ax_scatter = fig.add_subplot(2,3,2)
ax_scatter.scatter(x,y)

ax_bar = fig.add_subplot(2,3,3)
ax_bar.bar(x,y)

ax_bar = fig.add_subplot(2,3,4)
ax_bar.barh(x,y)

ax_pie = fig.add_subplot(2,3,5)
ax_pie.pie(y)

ax_hist = fig.add_subplot(2,3,6)
ax_hist.hist(freq)

fig.show()
```

```python
width = 6
height = 3

fig = plt.figure(figsize=(width,height))
ax = fig.subplots()

ax.plot(x,y, color='red', marker='o', label='Random value')
ax.plot(x,x, color='blue', marker='v', label='Original value')

ax.set_title('Example of Graph plotting')
ax.set_xlabel('Number')
ax.set_ylabel('Value')
ax.legend()

fig.show()

fig.savefig('mygraph.png')
```

**Plotting the same image with Two Y**

```python
fig = plt.figure(figsize=(10,5))
ax1 = fig.subplots()
ax2 = ax1.twinx()

thailand_df.plot(kind='line', x='date', y ='new_cases', ax = ax1)
ax1.legend(loc='upper left')
ax1.set_ylabel('Patient')
thailand_df.plot(kind='line', x='date', y ='new_deaths', ax = ax2, color='red')
ax2.legend(loc='upper right')
ax2.set_ylabel('Death')

fig.show()
```

**Summarize Situation in each contry**

```python
import datetime as dt
covid_df = pd.read_csv('/content/drive/MyDrive/owid-covid-data.csv', parse_dates=['date'])
dt.date.today()
```

```python
safe_date = dt.date.today() + pd.Timedelta('-2131 days')
safe_date
```

```python
recent_covid_df = covid_df[covid_df['date'] >= pd.Timestamp(safe_date)]
recent_covid_df
```

```python
locations = recent_covid_df.groupby('location')['new_cases'].sum() < 10
set(locations[locations].index)
```

```python
no_new_patients = set(locations[locations].index)
no_new_patients
```

**SD with Data analysis**

```python
locations = recent_covid_df.groupby('location')['new_cases'].std() < 10
low_sd = set(locations[locations].index)
low_sd
```

```python
safe = no_new_patients.union(low_sd)
```
